```python
import random

from jupyter_dash import JupyterDash 
# import dash

from dash import dcc, html, Input, Output, State
```

```python
def simulate_pop_up_pirate(num_slots):
    pop_slot = random.randint(1, num_slots)
    for turn in range(1, num_slots + 1):
        if turn == pop_slot:
            return 1 if turn % 2 else 2

app = JupyterDash(__name__)
accumulated_losses = {1: 0, 2: 0}

app.layout = html.Div([
    html.H1("Pirate Game Simulation 🏴‍☠️", style={
        'textAlign': 'center', 
        'color': '#22223b', 
        'marginBottom': '30px',
        'fontFamily': 'Trebuchet MS, sans-serif'
    }),
    html.Div([
        html.Div([
            html.Label("# of Barrel Slots:", style={'fontWeight': 'bold'}),
            dcc.Input(id='slots', value=20, type='number', min=2, style={
                'marginLeft': '10px', 'width': '80px'
            }),
        ], style={'marginBottom': '15px'}),
        html.Div([
            html.Label("Simulations:", style={'fontWeight': 'bold'}),
            dcc.Input(id='sims', value=100, type='number', min=1, style={
                'marginLeft': '10px', 'width': '80px'
            }),
        ], style={'marginBottom': '15px'}),
        html.Button('Run', id='run', style={
            'backgroundColor': '#4f772d',
            'color': 'white',
            'fontWeight': 'bold',
            'border': 'none',
            'padding': '8px 24px',
            'borderRadius': '6px',
            'fontSize': '16px',
            'cursor': 'pointer',
            'marginTop': '10px'
        }),
    ], style={
        'display': 'flex',
        'flexDirection': 'column',
        'alignItems': 'center',
        'justifyContent': 'center',
        'backgroundColor': '#f2e9e4',
        'padding': '24px',
        'borderRadius': '12px',
        'width': '350px',
        'margin': '0 auto'
    }),
    html.Div(id='batch-results', style={
        'marginTop': '30px',
        'textAlign': 'center',
        'fontSize': '18px',
        'color': '#1b263b'
    }, children="Click to simulate a batch"),
    html.Div(id='total-losses', style={
        'marginTop': '18px',
        'textAlign': 'center',
        'fontSize': '20px',
        'fontWeight': 'bold',
        'color': '#d90429'
    }, children="Total losses (whoever causes the pirate head to pop) will appear here.")
], style={
    'backgroundColor': '#e9ecef',
    'minHeight': '100vh',
    'padding': '20px'
})

@app.callback(
    [Output('batch-results', 'children'),
     Output('total-losses', 'children')],
    Input('run', 'n_clicks'),
    [State('slots', 'value'),
     State('sims', 'value')]
)
def update(n_clicks, slots, sims):
    if n_clicks is None or n_clicks == 0 or not slots or not sims:
        return ("Click to simulate a batch", "Total losses (whoever causes the pirate head to pop) will appear here.")
    
    batch = {1:0, 2:0}
    for _ in range(sims):
        loser = simulate_pop_up_pirate(slots)
        accumulated_losses[loser] += 1
        batch[loser] += 1
    
    return (
        f"Latest Batch: Player 1 - {batch[1]}, Player 2 - {batch[2]}",
        f"🏆 Total Losses: Player 1 - {accumulated_losses[1]}, Player 2 - {accumulated_losses[2]}"
    )

app.run_server(mode='inline')
```
